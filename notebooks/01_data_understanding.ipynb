# Data Understanding

```python
import pandas as pd

df_k1 = pd.read_csv('../data/raw/mamikos-2026-09-07-ukrida-kampus-1.csv')
df_k2 = pd.read_csv('../data/raw/mamikos-2026-09-09-ukrida-kampus-2.csv')

print(f"Ukuran Raw Kampus 1: {df_k1.shape}")
print(f"Ukuran Raw Kampus 2: {df_k2.shape}")
```

```python
display(df_k1.head(5))
```

```python
# Cek info tipe data dan unique values per kolom
data_summary = pd.DataFrame({
    'Data Type': df_k1.dtypes,
    'Null Count': df_k1.isnull().sum(),
    'Null Ratio (%)': (df_k1.isnull().sum() / len(df_k1) * 100).round(2),
    'Unique Count': df_k1.nunique()
})
display(data_summary)
```

```python
print(df_k1.isnull().sum())
```

```python
df_merged_temp = pd.concat([df_k1, df_k2], ignore_index=True)

total_raw = len(df_merged_temp)
duplikat_count = df_merged_temp.duplicated(subset=['nama_kost', 'harga']).sum()

print(f"Total Baris Gabungan Raw : {total_raw}")
print(f"Total Baris Duplikat     : {duplikat_count} ({round(duplikat_count/total_raw*100, 2)}%)")
print(f"Estimasi Data Unik       : {total_raw - duplikat_count}")
```

```python
# SEL 5: Inspeksi Sample Data Duplikat
df_merged_temp[df_merged_temp.duplicated(subset=['nama_kost', 'harga'], keep=False)].sort_values('nama_kost').head(10)
```

```python
df_tipe = df_merged_temp[df_merged_temp['nama_kost'].str.contains(r'(?i)\btipe\b', na=False)].copy()

df_tipe['suffix_tipe'] = df_tipe['nama_kost'].str.extract(r'(?i)\b(tipe\s+[a-z0-9\-_]+)', expand=False)

df_tipe['suffix_tipe'] = df_tipe['suffix_tipe'].str.title()

print(f"Total gabungan kost yang memiliki suffix tipe: {len(df_tipe)} baris\n")

print(df_tipe[['nama_kost', 'suffix_tipe', 'lokasi', 'harga']].head(15).to_string(index=False))

# Tampilkan sebaran tipe terbanyak
print("\nSebaran Suffix Tipe Terbanyak:")
print(df_tipe['suffix_tipe'].value_counts().head(10))
```

```python
# Konversi sederhana untuk memahami distribusi harga
df_merged_temp['harga_clean'] = (
    df_merged_temp['harga']
    .astype(str)
    .str.replace(r'[^\d]', '', regex=True)
)
df_merged_temp['harga_clean'] = pd.to_numeric(df_merged_temp['harga_clean'], errors='coerce')

# Ringkasan Statistik Harga
print("--- Ringkasan Statistik Harga Kost ---")
print(df_merged_temp['harga_clean'].describe())
```

```python
# Cek distribusi tipe gender kost (Putri/Putra/Campur)
print("--- Distribusi Tipe Kost ---")
print(df_merged_temp['tipe'].value_counts(dropna=False))

# Cek distribusi lokasi
print("\n--- Top 10 Lokasi Kost ---")
print(df_merged_temp['lokasi'].value_counts().head(10))
```
